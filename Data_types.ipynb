```python
fruits = ["apple" , "banana", "apple", "banana" ]
```

```python
print(fruits)
```

```python
print(fruits[0])
print(fruits[1])
```

```python
print(fruits[0])
print(fruits[-1])
```

Lists Method




```python
fruits.append("orange")  #add item
print(fruits)
```

```python
fruits.insert(1,"grapes") #insert the item at index 1 
print(fruits)
```

```python
fruits.remove("banana") #remove the item in the list
print(fruits)
```

```python
fruits.sort() #use to sort the list 
print(fruits)
```

```python
fruits.pop() #remove the last item in list
print(fruits)
```

Looping the list

```python
for fruit in fruits:
    print(fruit)
```

Tuple Methods

```python
numbers = (1,2,3,2,4,1)
print(numbers.count(2))
print(numbers.index(3))
```

SET (they store unique unordered values)

```python
my_set = {1,2,3,4,5,5}
print(my_set)
```

Set Methods 

```python
my_set.add(6)
my_set.remove(2)
print(my_set)
```

Set Operations

```python
A = {1,2,3}
B = {3,4,5}

print(A.union(B))
print(A.intersection(B))
print(A.difference(B))
```

Dictionaries (Stores Data as key : value pairs)

```python
Student = {
    "name" : "ali",
    "age" : 23,
    "course" : "software Engineering",
    "CGPA" : 3.87
}

print(Student)
```

Accessing Elements 

```python
print(Student["name"])
print(Student.get("age"))
```

Dictionaries Methods 

```python
Student["name"] = "hammad"
```

```python
Student["age"] = 24
```

```python
Student.pop("course")
```

```python
print(Student)
```

Looping Dictionary 

```python
for key, value in Student.items():
    print(key, ":", value)
```
